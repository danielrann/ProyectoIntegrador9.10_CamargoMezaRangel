## Definición de la variable objetivo

La variable objetivo del problema se construye a partir de la columna **loan_status**.

Se define la variable binaria **default** de la siguiente forma:

- **0 = Fully Paid**
- **1 = Charged Off**

Sin embargo, la columna **loan_status** contiene también otros estados intermedios como **Current**, **Late (31-120 days)** o **In Grace Period**. Estos estados no representan un desenlace final del préstamo, pero para el proposito de la evaluación de los modelos a usar, se tomarán en cuenta para el valor de **Fully paid**.


Después de esta categorización, se genera la variable **default**.

```python
df["default"] = df["loan_status"].apply(lambda x: 1 if x == "Charged Off" else 0)
```

Podemos ver los resultados de la clasificación de la siguiente forma:

```python
print("Distribución de la variable default:")
print(df["default"].value_counts(dropna=False))
print("Distribución porcentual de la variable default:")
print((df["default"].value_counts(normalize=True) * 100).round(2))
df[["loan_status", "default"]].head(15)
```
