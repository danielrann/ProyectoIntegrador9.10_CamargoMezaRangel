# **Resumen Ejecutivo Del EDA**

En esta sección se sintetizan los principales hallazgos obtenidos durante el análisis exploratorio de datos, con el fin de resumir la calidad del dataset, identificar las variables más prometedoras para la predicción de default y establecer decisiones preliminares que guiarán el preprocesamiento y modelado posteriores.

## **Síntesis de hallazgos del EDA**

| Aspecto | Resumen |
|---|---|
| **Calidad de los datos** | El dataset presenta una alta dimensionalidad, con predominio de variables numéricas y presencia de valores faltantes en múltiples columnas. Para el análisis se aplicó un criterio de completitud, conservando variables con un nivel de cobertura adecuado. |
| **Variables más prometedoras para predecir default** | Entre las variables numéricas, `int_rate` y `fico_range_high` mostraron la asociación más clara con `default`, seguidas por `revol_util` y `dti`. Entre las variables categóricas, `purpose`, `home_ownership`, `addr_state` y `emp_length` mostraron diferencias relevantes en sus tasas de impago. |
| **Problemas detectados** | Se identificó desbalance de clases en la variable objetivo (`default`), presencia de outliers severos en variables como `annual_inc` y `dti`, correlación relativamente alta entre `open_acc` y `total_acc`, y valores faltantes sistemáticos en varias variables del dataset original. |
| **Decisiones preliminares para el preprocesamiento** | Se mantendrán las variables seleccionadas por relevancia y calidad de datos. Se considerará tratamiento de nulos según el tipo de variable, posible transformación de variables altamente sesgadas como `annual_inc`, vigilancia de redundancias entre variables correlacionadas y uso de partición estratificada y métricas robustas debido al desbalance de clases. |

En conjunto, el EDA permitió identificar un subconjunto de variables con potencial predictivo para el problema de clasificación de default, así como varios desafíos propios de un dataset financiero real, tales como valores faltantes, outliers, asimetría y desbalance de clases. Estos hallazgos servirán como base para definir la estrategia de preprocesamiento y construir posteriormente los modelos en scikit-learn y PySpark.
